```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Loading dataset
df = pd.read_csv("data/Unemployment in India.csv")

# Display first 5 rows
df.head()
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nTotal Missing Values:", df.isnull().sum().sum())
```

```python
print("Total Duplicate Rows:", df.duplicated().sum())
```

```python
# Checking duplicate records

print("Total Duplicate Rows:", df.duplicated().sum())
```

```python
# Removing extra spaces from column names

df.columns = df.columns.str.strip()

# Display cleaned column names

print("Cleaned Column Names:")
print(df.columns.tolist())
```

```python
# Checking completely empty rows

empty_rows = df.isnull().all(axis=1).sum()

print("Completely Empty Rows:", empty_rows)
```

```python
# Removing completely empty rows

df = df.dropna(how='all')

# Resetting index

df = df.reset_index(drop=True)

print("Dataset Shape After Removing Empty Rows:", df.shape)

# Checking missing values again

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Converting Date column into datetime format

df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)

# Checking updated data types

print(df.dtypes)
```

```python
# Sorting dataset by date

df = df.sort_values(by='Date')

# Resetting index

df = df.reset_index(drop=True)

# Display first 5 rows

df.head()
```

```python
print("Final Dataset Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# Statistical summary of the dataset

df.describe()
```

```python
# Checking unique regions

print("Total Regions:", df['Region'].nunique())

print("\nList of Regions:")
print(sorted(df['Region'].unique()))
```

```python
# Checking frequency of data

print(df['Frequency'].value_counts())

print("\nArea Distribution:")
print(df['Area'].value_counts())
```


## 1. State-wise Average Unemployment Rate

This analysis calculates the average unemployment rate across different Indian states and regions. It helps identify areas with comparatively higher and lower unemployment rates.


```python

# Calculating average unemployment rate for each region

region_avg = df.groupby('Region')['Estimated Unemployment Rate (%)'].mean().sort_values(ascending=False)

print("Average Unemployment Rate by Region:")
display(region_avg)
```

```python

plt.figure(figsize=(14, 7))

region_avg.plot(kind='bar', color='skyblue', edgecolor='black')

plt.title('Average Unemployment Rate by Indian States/Regions')
plt.xlabel('States / Regions')
plt.ylabel('Average Unemployment Rate (%)')

plt.xticks(rotation=90)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# Visualizing average unemployment rate by region

plt.figure(figsize=(14, 7))

region_avg.plot(
    kind='bar',
    color='skyblue',
    edgecolor='black'
)

plt.title('Average Unemployment Rate by Indian States/Regions',
          fontsize=16)

plt.xlabel('States / Regions', fontsize=12)
plt.ylabel('Average Unemployment Rate (%)', fontsize=12)

plt.xticks(rotation=90)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```


### Observation

The bar chart shows the average unemployment rate across different Indian states and regions. It highlights regional differences in unemployment levels. Some regions have higher average unemployment rates, while others have comparatively lower rates.



## 2. Month-wise Unemployment Trend

This analysis examines how the unemployment rate changes over time in India. It helps identify monthly fluctuations and understand employment trends.


```python

# Calculating average unemployment rate month-wise

monthly_avg = df.groupby(
    df['Date'].dt.to_period('M')
)['Estimated Unemployment Rate (%)'].mean()

print(monthly_avg)
```

```python

# Plotting monthly unemployment trend

plt.figure(figsize=(14, 6))

monthly_avg.index = monthly_avg.index.astype(str)

plt.plot(monthly_avg.index, monthly_avg.values,
         marker='o', color='blue', linewidth=2)

plt.title('Monthly Unemployment Rate Trend in India', fontsize=16)
plt.xlabel('Month')
plt.ylabel('Average Unemployment Rate (%)')

plt.xticks(rotation=90)
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

### Observation

The monthly trend line chart shows fluctuations in unemployment rates over time. It helps identify periods of increasing and decreasing unemployment and provides a basis for studying the impact of the COVID-19 pandemic.

## 3. COVID-19 Impact on Unemployment in India

This analysis compares unemployment rates before and after the COVID-19 pandemic to understand how the pandemic affected employment conditions in India.

```python
# Creating pre-COVID and post-COVID groups

pre_covid = df[df['Date'] < '2020-03-01']

post_covid = df[df['Date'] >= '2020-03-01']

# Calculating average unemployment rates

pre_covid_avg = pre_covid['Estimated Unemployment Rate (%)'].mean()

post_covid_avg = post_covid['Estimated Unemployment Rate (%)'].mean()

print("Pre-COVID Average Unemployment Rate:",
      round(pre_covid_avg, 2), "%")

print("Post-COVID Average Unemployment Rate:",
      round(post_covid_avg, 2), "%")
```

```python
# Comparing pre-COVID and post-COVID unemployment rates

covid_comparison = pd.Series({
    'Pre-COVID': pre_covid_avg,
    'Post-COVID': post_covid_avg
})

plt.figure(figsize=(8, 5))

covid_comparison.plot(
    kind='bar',
    color=['skyblue', 'salmon'],
    edgecolor='black'
)

plt.title('Pre-COVID vs Post-COVID Unemployment Rate')
plt.xlabel('Period')
plt.ylabel('Average Unemployment Rate (%)')

plt.xticks(rotation=0)
plt.grid(axis='y', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

## 4. Unemployment Trends in Three Major States

This analysis compares unemployment rate trends over time in Maharashtra, Haryana, and Uttar Pradesh.

```python
# Selecting three states
selected_states = ['Maharashtra', 'Haryana', 'Uttar Pradesh']

# Filtering the dataset
state_data = df[df['Region'].isin(selected_states)]

# Grouping data by date and state
state_trend = state_data.groupby(
    ['Date', 'Region']
)['Estimated Unemployment Rate (%)'].mean().reset_index()

# Plotting the time-series chart
plt.figure(figsize=(14, 7))

for state in selected_states:
    state_df = state_trend[state_trend['Region'] == state]
    plt.plot(
        state_df['Date'],
        state_df['Estimated Unemployment Rate (%)'],
        marker='o',
        linewidth=2,
        label=state
    )

plt.title('Unemployment Rate Trends in Three Indian States')
plt.xlabel('Date')
plt.ylabel('Unemployment Rate (%)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

**Observation:**

The line chart compares unemployment trends in Maharashtra, Haryana, and Uttar Pradesh. It shows how unemployment rates changed over time in these states. The trends help identify differences in regional employment conditions.

## 5. Top 10 States with Highest Average Unemployment Rate

This analysis identifies the top 10 Indian states and regions with the highest average unemployment rates.

```python
# Selecting the top 10 regions
top_10_regions = region_avg.head(10)

# Creating the bar chart
plt.figure(figsize=(12, 6))

top_10_regions.plot(
    kind='bar',
    color='orange',
    edgecolor='black'
)

plt.title('Top 10 States with Highest Average Unemployment Rate')
plt.xlabel('States / Regions')
plt.ylabel('Average Unemployment Rate (%)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

## 6. Correlation Heatmap

A correlation heatmap helps understand the relationship between unemployment rate, estimated employment, and labour participation rate.

```python
# Selecting numerical columns
correlation_data = df[
    [
        'Estimated Unemployment Rate (%)',
        'Estimated Employed',
        'Estimated Labour Participation Rate (%)'
    ]
]

# Calculating correlation
correlation_matrix = correlation_data.corr()

# Creating heatmap
plt.figure(figsize=(9, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Unemployment and Employment Indicators')
plt.tight_layout()
plt.show()
```
